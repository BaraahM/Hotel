#**2. Data Preprocessing**

## 2.1 Handling Missing Values:

```python
# Check missing values before handling
print("--- Missing Values Count Before Imputation ---")
print(df[['company', 'agent', 'country', 'children']].isnull().sum())

# Impute 'agent' and 'company' missing values with 0 (representing 'None')
df['agent'] = df['agent'].fillna(0)
df['company'] = df['company'].fillna(0)

# 2. Impute 'children' missing values with 0
df['children'] = df['children'].fillna(0)

# 3. Impute 'country' missing values with a placeholder 'Unknown'
df['country'] = df['country'].fillna('Unknown')

# Verify that missing values have been successfully addressed
print("\n--- Missing Values Count After Imputation ---")
print(df[['company', 'agent', 'country', 'children']].isnull().sum())
```

## 2.2 Addressing Data Inconsistencies

To ensure logical validity, we clean the dataset of operational contradictions:
1. **Zero Total Guests**: Bookings where `adults`, `children`, and `babies` are all `0` are logically impossible and represent errors or cancellations before entry.
2. **Negative ADR**: The Average Daily Rate (`adr`) cannot be negative. Any row with a negative ADR is an anomaly and will be removed.

```python
# Initial shape assessment
initial_rows = df.shape[0]
print(f"Initial number of records: {initial_rows}")

# 1. Filter out rows where the total sum of guests is 0
zero_guests_condition = (df['adults'] == 0) & (df['children'] == 0) & (df['babies'] == 0)
df = df[~zero_guests_condition]
print(f"Records after removing zero-guest rows: {df.shape[0]}")

# 2. Filter out rows where ADR is negative
df = df[df['adr'] >= 0]
print(f"Final records after dropping negative ADR rows: {df.shape[0]}")

# Calculate total rows dropped
print(f"Total operational anomaly rows removed: {initial_rows - df.shape[0]}")
```

## 2.3 Feature Engineering

To provide richer insights for our EDA and predictive model, we generate three engineered metrics:
1. **`total_guests`**: Combining `adults` + `children` + `babies` captures the true demographic load of a booking.
2. **`total_stay_duration`**: Combining weekend stays and weekday stays allows us to analyze full lengths of stay uniformly.
3. **`arrival_date`**: Combining `arrival_date_year`, `arrival_date_month`, and `arrival_date_day_of_month` into a structured datetime object to unlock proper time-series calculations and seamless visualization sorting.

```python
# 1. Create total_guests feature
df['total_guests'] = df['adults'] + df['children'] + df['babies']

# 2. Create total_stay_duration feature
df['total_stay_duration'] = df['stays_in_weekend_nights'] + df['stays_in_week_nights']

# 3. Construct a standard datetime object for arrival date
# Map textual month names to standard numeric values
month_mapping = {
    'January': 1, 'February': 2, 'March': 3, 'April': 4,
    'May': 5, 'June': 6, 'July': 7, 'August': 8,
    'September': 9, 'October': 10, 'November': 11, 'December': 12
}
df['arrival_date_month_num'] = df['arrival_date_month'].map(month_mapping)

# Use pandas to_datetime to build the clean timestamp feature
df['arrival_date'] = pd.to_datetime(
    df['arrival_date_year'].astype(str) + '-' +
    df['arrival_date_month_num'].astype(str) + '-' +
    df['arrival_date_day_of_month'].astype(str)
)

# Safely drop our temporary month number column
df.drop(columns=['arrival_date_month_num'], inplace=True)

# Display engineered additions to ensure valid formatting
df[['total_guests', 'total_stay_duration', 'arrival_date']].head()
```

## 2.4 Categorical Encoding

Machine Learning algorithms require numerical matrix inputs. We handle our nominal attributes using modern standard encoding methodologies:
* **Binary Encoding**: For the `hotel` column, we assign a boolean integer (`Resort Hotel` $\rightarrow$ 0, `City Hotel` $\rightarrow$ 1).
* **One-Hot Encoding**: Multi-class string categories (`meal`, `market_segment`, `deposit_type`) are expanded into multiple dummy indicator variables using `drop_first=True` to avoid structural multicollinearity (the dummy variable trap).

```python
# 1. Map binary string feature 'hotel' to numeric values
df['hotel'] = df['hotel'].map({'Resort Hotel': 0, 'City Hotel': 1})

# List of nominal categorical attributes to scale out via one-hot encoding
categorical_columns = ['meal', 'market_segment', 'deposit_type']

# 2. Execute One-Hot Encoding using pd.get_dummies
df = pd.get_dummies(df, columns=categorical_columns, drop_first=True)

# preventing the linear dependence model trap.
print("Columns after structural transformation:")
print(df.columns.tolist()[:15])
```

## 2.5 Feature Scaling

Numerical variables such as `lead_time` and `adr` exhibit vastly different ranges and structural right-skewness. Leaving them in their raw ranges could heavily bias distance-based or optimization-heavy ML estimators. We perform **Z-score Standardization** via `StandardScaler`, centering the features at a mean of 0 and scaling them to unit variance.

```python
from sklearn.preprocessing import StandardScaler

# Instantiate standard scaling transformation engine
scaler = StandardScaler()

# Target skewed operational continuous columns
scaling_targets = ['lead_time', 'adr']

# Fit scaler parameters and map standardized shifts back to the dataframe
df[scaling_targets] = scaler.fit_transform(df[scaling_targets])

# Validate results by inspecting transformed statistical configurations
print("Transformed Numerical Features Summary:")
df[scaling_targets].describe().round(3)
```
